```
using System;

class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; set; }
    public string BankName { get; set; }
    public string CardType { get; set; }
    public bool IsBlocked { get; set; }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal balance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;

        BankName = "ТБанк";
        CardType = "Обычная";
        IsBlocked = false;
    }

    public virtual string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" + $"Холдер: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.\n" +
               $"Банк: {BankName}\n" +
               $"Тип карты: {CardType}\n" +
               $"Заблокирована: {IsBlocked}";
    }

    public virtual void Pay(decimal amount)
    {
        if (amount <= Balance && !IsBlocked)
        {
            Balance -= amount;
            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств или карта заблокирована.");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Баланс карты: {Balance} руб.");
    }

    public void Transfer(CreditCard otherCard, decimal amount)
    {
        if (amount <= Balance && !IsBlocked)
        {
            Balance -= amount;
            otherCard.Balance += amount;

            Console.WriteLine(
                $"{HolderName} перевел {amount} руб. пользователю {otherCard.HolderName}."
            );
        }
        else
        {
            Console.WriteLine("Недостаточно средств или карта заблокирована.");
        }
    }

    public void BlockCard()
    {
        IsBlocked = true;
        Console.WriteLine("Карта заблокирована.");
    }

    public void UnblockCard()
    {
        IsBlocked = false;
        Console.WriteLine("Карта разблокирована.");
    }

    public void ChangeBank(string newBank)
    {
        BankName = newBank;
        Console.WriteLine($"Банк изменен на: {BankName}");
    }
}

class GoldCreditCard : CreditCard
{
    public int BonusMiles { get; set; }
    public int BonusLevel { get; set; }
    public decimal CashbackPercent { get; set; }

    public GoldCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        BonusMiles = bonusMiles;
        BonusLevel = 1;
        CashbackPercent = 2;
        CardType = "Gold";
    }

    public override void Pay(decimal amount)
    {
        if (amount <= Balance)
        {
            Balance -= amount;

            int miles = (int)(amount / 100);
            BonusMiles += miles;

            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
            Console.WriteLine($"Начислено бонусных миль: {miles}.");
            Console.WriteLine($"Всего бонусных миль: {BonusMiles}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств для оплаты.");
        }
    }

    public void UseBonusMiles(int miles)
    {
        if (miles <= BonusMiles)
        {
            BonusMiles -= miles;
            Console.WriteLine($"Использовано бонусных миль: {miles}.");
        }
        else
        {
            Console.WriteLine("Недостаточно бонусных миль.");
        }
    }

    public decimal CalculateCashback(decimal amount)
    {
        return amount * CashbackPercent / 100;
    }

    public void ShowBonusInfo()
    {
        Console.WriteLine($"Бонусный уровень: {BonusLevel}");
        Console.WriteLine($"Бонусные мили: {BonusMiles}");
        Console.WriteLine($"Кэшбэк: {CashbackPercent}%");
    }
}


class PremiumCreditCard : CreditCard
{
    public string SupportAssistant { get; set; }
    public bool PriorityService { get; set; }
    public bool Insurance { get; set; }

    public PremiumCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        SupportAssistant = supportAssistant;
        PriorityService = true;
        Insurance = true;
        CardType = "Premium";
    }

    public override void CheckBalance()
    {
        if (Balance >= 0)
        {
            Console.WriteLine($"Баланс карты: {Balance} руб.");
        }
        else
        {
            Console.WriteLine("Обнаружена проблема с балансом.");
            Console.WriteLine($"Обратитесь к ассистенту поддержки: {SupportAssistant}.");
        }
    }

    public void RequestPriorityService()
    {
        if (PriorityService)
        {
            Console.WriteLine("Запрошено приоритетное обслуживание.");
        }
    }

    public void ShowInsurance()
    {
        if (Insurance)
        {
            Console.WriteLine("Страховка карты подключена.");
        }
        else
        {
            Console.WriteLine("Страховка карты отсутствует.");
        }
    }

    public void ChangeAssistant(string newAssistant)
    {
        SupportAssistant = newAssistant;
        Console.WriteLine($"Новый ассистент: {SupportAssistant}");
    }
}


interface IOnlinePayment
{
    void OnlinePayment(decimal amount);
}

interface ITransfer
{
    void MakeTransfer(CreditCard card, decimal amount);
}


class CorporateCreditCard : CreditCard, IOnlinePayment, ITransfer
{
    public string Company { get; set; }
    public string CompanyPosition { get; set; }
    public decimal CompanyLimit { get; set; }
    public int EmployeeCount { get; set; }

    public CorporateCreditCard(
        string cardNumber,
        string holderName,
        string expiryDate,
        decimal balance,
        string company)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        Company = company;
        CompanyPosition = "Сотрудник";
        CompanyLimit = 500000;
        EmployeeCount = 1;
        CardType = "Corporate";
    }

    public override string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" +
               $"Холдер: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.\n" +
               $"Компания: {Company}\n" +
               $"Должность: {CompanyPosition}\n" +
               $"Лимит компании: {CompanyLimit} руб.\n" +
               $"Количество сотрудников: {EmployeeCount}";
    }

    public void IncreaseLimit(decimal amount)
    {
        CompanyLimit += amount;
        Console.WriteLine($"Лимит увеличен до {CompanyLimit} руб.");
    }

    public void ShowCompanyInfo()
    {
        Console.WriteLine($"Компания: {Company}");
        Console.WriteLine($"Должность: {CompanyPosition}");
        Console.WriteLine($"Сотрудников: {EmployeeCount}");
    }

    public void AddEmployee()
    {
        EmployeeCount++;
        Console.WriteLine($"Количество сотрудников: {EmployeeCount}");
    }

    public void OnlinePayment(decimal amount)
    {
        Pay(amount);
    }

    public void MakeTransfer(CreditCard card, decimal amount)
    {
        Transfer(card, amount);
    }
}


class PremiumGoldCreditCard : GoldCreditCard
{
    public string PersonalManager { get; set; }
    public decimal SpecialLimit { get; set; }

    public PremiumGoldCreditCard(
        string cardNumber,
        string holderName,
        string expiryDate,
        decimal balance,
        int bonusMiles,
        string personalManager)
        : base(cardNumber, holderName, expiryDate, balance, bonusMiles)
    {
        PersonalManager = personalManager;
        SpecialLimit = 1000000;
        CardType = "Premium Gold";
    }

    public void ShowManager()
    {
        Console.WriteLine($"Персональный менеджер: {PersonalManager}");
    }

    public void IncreaseSpecialLimit(decimal amount)
    {
        SpecialLimit += amount;
        Console.WriteLine($"Специальный лимит: {SpecialLimit} руб.");
    }

    public void ContactManager()
    {
        Console.WriteLine($"Связь с персональным менеджером: {PersonalManager}");
    }
}


CreditCard card = new CreditCard("1111-2222-3333-4444", "Иван Иванов", "28.12.2036", 50000);
GoldCreditCard goldCard = new GoldCreditCard("2222-3333-4444-5555", "Петр Петров", "29.06.2030", 100000, 500);
PremiumCreditCard premiumCard = new PremiumCreditCard("3333-4444-5555-6666", "Анна Смирнова", "30.07.2029", 75000, "Анна");
CorporateCreditCard corporateCard = new CorporateCreditCard("4444-5555-6666-7777", "Сергей Сидоров", "18.09.2028", 200000, "ООО Ромашка");
PremiumGoldCreditCard premiumGoldCard = new PremiumGoldCreditCard("5555-6666-7777-8888", "Мария Петрова", "15.05.2032", 150000, 1000,
    "Елена");

Console.WriteLine("Обычная кредитная карта");
Console.WriteLine(card.GetInfo());
card.CheckBalance();
card.ChangeBank("Сбербанк");

Console.WriteLine();

Console.WriteLine("Золотая кредитная карта");
Console.WriteLine(goldCard.GetInfo());
goldCard.Pay(10000);
goldCard.ShowBonusInfo();
Console.WriteLine($"Кэшбэк с 10000 руб.: {goldCard.CalculateCashback(10000)} руб.");

Console.WriteLine();

Console.WriteLine("Премиальная кредитная карта");
Console.WriteLine(premiumCard.GetInfo());
premiumCard.CheckBalance();
premiumCard.RequestPriorityService();
premiumCard.ShowInsurance();

Console.WriteLine();

Console.WriteLine("Корпоративная кредитная карта");
Console.WriteLine(corporateCard.GetInfo());
corporateCard.Pay(15000);
corporateCard.ShowCompanyInfo();
corporateCard.IncreaseLimit(100000);
corporateCard.AddEmployee();

Console.WriteLine();

Console.WriteLine("Многоуровневое наследование");
Console.WriteLine(premiumGoldCard.GetInfo());
premiumGoldCard.ShowBonusInfo();
premiumGoldCard.ShowManager();
premiumGoldCard.IncreaseSpecialLimit(200000);

Console.WriteLine();

Console.WriteLine("Множественное наследование через интерфейсы");

corporateCard.OnlinePayment(5000);
corporateCard.MakeTransfer(premiumCard, 3000);

Console.WriteLine();

Console.WriteLine("Взаимодействие объектов");

Console.WriteLine($"Баланс Петра до перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны до перевода: {premiumCard.Balance} руб.");

goldCard.Transfer(premiumCard, 5000);

Console.WriteLine($"Баланс Петра после перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны после перевода: {premiumCard.Balance} руб.");

Console.WriteLine();

CreditCard selectedCard;

selectedCard = goldCard;
selectedCard.Pay(1000);

selectedCard = premiumCard;
selectedCard.CheckBalance();

selectedCard = corporateCard;
Console.WriteLine(selectedCard.GetInfo());
```
