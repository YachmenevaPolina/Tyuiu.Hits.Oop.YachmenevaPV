<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; }
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; }
}


public class Supplier : ISupplier
{
    public string SupplierName { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; set; }

    public Supplier(string supplierName, decimal cost, string supplierType)
    {
        SupplierName = supplierName;
        Cost = cost;
        SupplierType = supplierType;
    }

    public string GetName()
    {
        return SupplierName;
    }

    public decimal GetCost()
    {
        return Cost;
    }
}

public class Employee : IEmployee
{
    public string FullName { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }

    public Employee(string fullName, string position, decimal salary)
    {
        FullName = fullName;
        Position = position;
        Salary = salary;
    }

    public string GetFullName()
    {
        return FullName;
    }

    public decimal GetSalary()
    {
        return Salary;
    }
}


public class Enterprise : IEnterprise
{
    private List<ISupplier> suppliers;
    private List<IEmployee> employees;

    public string Name { get; set; }

    public Enterprise(string name)
    {
        Name = name;
        suppliers = new List<ISupplier>();
        employees = new List<IEmployee>();
    }

    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    // Получение списка поставщиков
    public List<ISupplier> GetSuppliers()
    {
        return suppliers;
    }

    public List<IEmployee> GetEmployees()
    {
        return employees;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal totalExpenses = 0;

        foreach (ISupplier supplier in suppliers)
        {
            totalExpenses += supplier.GetCost();
        }

        foreach (IEmployee employee in employees)
        {
            totalExpenses += employee.GetSalary();
        }

        return totalExpenses;
    }
}


public class EquipmentSupplier : ISupplier
{
    public string SupplierName { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; } = "Оборудование";

    public EquipmentSupplier(string supplierName, decimal cost)
    {
        SupplierName = supplierName;
        Cost = cost;
    }

    public string GetName()
    {
        return SupplierName;
    }

    public decimal GetCost()
    {
        return Cost;
    }
}


public class Manager : IEmployee
{
    public string FullName { get; set; }
    public string Position { get; } = "Руководитель";
    public decimal Salary { get; set; }

    public Manager(string fullName, decimal salary)
    {
        FullName = fullName;
        Salary = salary;
    }

    public string GetFullName()
    {
        return FullName;
    }

    public decimal GetSalary()
    {
        return Salary;
    }
}


Enterprise enterprise = new Enterprise("ТехноПром");

ISupplier supplier1 = new Supplier("Логистический центр", 100000, "Логистика");

ISupplier supplier2 = new Supplier("Поставщик офисной мебели", 50000, "Офисное оборудование");

ISupplier supplier3 = new EquipmentSupplier("Поставщик компьютерного оборудования", 75000);

enterprise.AddSupplier(supplier1);
enterprise.AddSupplier(supplier2);
enterprise.AddSupplier(supplier3);

IEmployee employee1 = new Employee("Иван Иванов", "Менеджер", 70000);

IEmployee employee2 = new Employee("Анна Петрова", "Разработчик", 95000);

IEmployee employee3 = new Manager("Сергей Смирнов", 120000);

enterprise.AddEmployee(employee1);
enterprise.AddEmployee(employee2);
enterprise.AddEmployee(employee3);

Console.WriteLine("========================================");
Console.WriteLine("       ПРЕДПРИЯТИЕ: " + enterprise.Name);
Console.WriteLine("========================================");

Console.WriteLine("\nСПИСОК ПОСТАВЩИКОВ:");

int supplierNumber = 1;

foreach (ISupplier supplier in enterprise.GetSuppliers())
{
    Console.WriteLine($"{supplierNumber}. {supplier.GetName()}");
    Console.WriteLine($"   Тип поставщика: {supplier.SupplierType}");
    Console.WriteLine($"   Стоимость услуг: {supplier.GetCost():N0} руб.");

    supplierNumber++;
}

Console.WriteLine("\nСПИСОК СОТРУДНИКОВ:");

int employeeNumber = 1;

foreach (IEmployee employee in enterprise.GetEmployees())
{
    Console.WriteLine($"{employeeNumber}. {employee.GetFullName()}");
    Console.WriteLine($"   Должность: {employee.Position}");
    Console.WriteLine($"   Зарплата: {employee.GetSalary():N0} руб.");

    employeeNumber++;
}

Console.WriteLine("========================================");
Console.WriteLine(
    $"ОБЩИЕ РАСХОДЫ ПРЕДПРИЯТИЯ: {enterprise.CalculateTotalExpenses():N0} руб."
);
Console.WriteLine("========================================");
```
