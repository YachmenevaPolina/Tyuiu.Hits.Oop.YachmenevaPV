```
using System;
using System.Collections.Generic;

interface IPayment
{
    void Pay(decimal amount);
}

interface ICardInfo
{
    string GetInfo();
}

interface IOnlinePayment
{
    void OnlinePayment(decimal amount);
}

interface ITransfer
{
    void MakeTransfer(CreditCard card, decimal amount);
}

class CreditCard : IPayment, ICardInfo
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; set; }
    public string BankName { get; set; }
    public string CardType { get; set; }
    public bool IsBlocked { get; set; }
    public string PaymentSystem { get; set; }
    public int DailyLimit { get; set; }
    public int PinAttempts { get; set; }
    public DateTime LastOperationDate { get; set; }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal balance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;

        BankName = "ТБанк";
        CardType = "Обычная";
        IsBlocked = false;
        PaymentSystem = "МИР";
        DailyLimit = 100000;
        PinAttempts = 0;
        LastOperationDate = DateTime.Now;
    }

    public virtual string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" +
               $"Владелец: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.\n" +
               $"Банк: {BankName}\n" +
               $"Тип карты: {CardType}\n" +
               $"Платежная система: {PaymentSystem}\n" +
               $"Дневной лимит: {DailyLimit} руб.\n" +
               $"Количество попыток ввода PIN: {PinAttempts}\n" +
               $"Заблокирована: {IsBlocked}";
    }

    public virtual void Pay(decimal amount)
    {
        if (amount > 0 && amount <= Balance && !IsBlocked && amount <= DailyLimit)
        {
            Balance -= amount;
            LastOperationDate = DateTime.Now;
            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств, неверная сумма, превышен лимит или карта заблокирована.");
        }
    }

    void IPayment.Pay(decimal amount)
    {
        Pay(amount);
    }

    string ICardInfo.GetInfo()
    {
        return GetInfo();
    }

    public void Pay(decimal amount, string description)
    {
        Console.WriteLine($"Назначение платежа: {description}");
        Pay(amount);
    }

    public void Pay(decimal amount, string description, string category)
    {
        Console.WriteLine($"Назначение платежа: {description}");
        Console.WriteLine($"Категория: {category}");
        Pay(amount);
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Баланс карты: {Balance} руб.");
    }

    public void Transfer(CreditCard otherCard, decimal amount)
    {
        if (otherCard != null && otherCard != this &&
            amount > 0 && amount <= Balance && !IsBlocked &&
            !otherCard.IsBlocked && amount <= DailyLimit)
        {
            Balance -= amount;
            otherCard.Balance += amount;

            LastOperationDate = DateTime.Now;
            otherCard.LastOperationDate = DateTime.Now;

            Console.WriteLine(
                $"{HolderName} перевел {amount} руб. пользователю {otherCard.HolderName}."
            );
        }
        else
        {
            Console.WriteLine("Перевод невозможен. Проверьте сумму и состояние карт.");
        }
    }

    public void BlockCard()
    {
        IsBlocked = true;
        Console.WriteLine("Карта заблокирована.");
    }

    public void UnblockCard()
    {
        IsBlocked = false;
        Console.WriteLine("Карта разблокирована.");
    }

    public void ChangeBank(string newBank)
    {
        BankName = newBank;
        Console.WriteLine($"Банк изменен на: {BankName}");
    }

    public void ChangePaymentSystem(string system)
    {
        PaymentSystem = system;
        Console.WriteLine($"Платежная система изменена на: {PaymentSystem}");
    }

    public void ShowLastOperation()
    {
        Console.WriteLine($"Последняя операция: {LastOperationDate}");
    }

    public void ChangeDailyLimit(int limit)
    {
        if (limit > 0)
        {
            DailyLimit = limit;
            Console.WriteLine($"Дневной лимит изменен: {DailyLimit} руб.");
        }
        else
        {
            Console.WriteLine("Лимит должен быть больше нуля.");
        }
    }

    public void RegisterPinAttempt()
    {
        PinAttempts++;

        if (PinAttempts >= 3)
        {
            BlockCard();
            Console.WriteLine("Превышено количество попыток ввода PIN.");
        }
        else
        {
            Console.WriteLine($"Количество попыток ввода PIN: {PinAttempts}");
        }
    }
}

class GoldCreditCard : CreditCard
{
    public int BonusMiles { get; set; }
    public int BonusLevel { get; set; }
    public decimal CashbackPercent { get; set; }
    public int BonusPoints { get; set; }
    public decimal MaximumCashback { get; set; }
    public string BonusProgram { get; set; }

    public GoldCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        BonusMiles = bonusMiles;
        BonusLevel = 1;
        CashbackPercent = 2;
        CardType = "Gold";
        BonusPoints = 0;
        MaximumCashback = 5000;
        BonusProgram = "Gold Bonus";
    }

    public override void Pay(decimal amount)
    {
        if (amount > 0 && amount <= Balance && !IsBlocked && amount <= DailyLimit)
        {
            Balance -= amount;

            int miles = (int)(amount / 100);
            BonusMiles += miles;

            LastOperationDate = DateTime.Now;

            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
            Console.WriteLine($"Начислено бонусных миль: {miles}.");
            Console.WriteLine($"Всего бонусных миль: {BonusMiles}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств, неверная сумма, превышен лимит или карта заблокирована.");
        }
    }

    public void UseBonusMiles(int miles)
    {
        if (miles > 0 && miles <= BonusMiles)
        {
            BonusMiles -= miles;
            Console.WriteLine($"Использовано бонусных миль: {miles}.");
        }
        else
        {
            Console.WriteLine("Недостаточно бонусных миль или неверное количество.");
        }
    }

    public decimal CalculateCashback(decimal amount)
    {
        decimal cashback = amount * CashbackPercent / 100;

        if (cashback > MaximumCashback)
        {
            cashback = MaximumCashback;
        }

        return cashback;
    }

    public void ShowBonusInfo()
    {
        Console.WriteLine($"Бонусный уровень: {BonusLevel}");
        Console.WriteLine($"Бонусные мили: {BonusMiles}");
        Console.WriteLine($"Кэшбэк: {CashbackPercent}%");
    }

    public void AddBonusPoints(int points)
    {
        if (points > 0)
        {
            BonusPoints += points;
            Console.WriteLine($"Всего бонусных баллов: {BonusPoints}");
        }
    }

    public void ShowBonusProgram()
    {
        Console.WriteLine($"Программа: {BonusProgram}");
        Console.WriteLine($"Максимальный кэшбэк: {MaximumCashback} руб.");
    }

    public void UpgradeBonusLevel()
    {
        BonusLevel++;
        CashbackPercent += 0.5m;
        Console.WriteLine($"Новый бонусный уровень: {BonusLevel}");
        Console.WriteLine($"Новый кэшбэк: {CashbackPercent}%");
    }
}

class PremiumCreditCard : CreditCard
{
    public string SupportAssistant { get; set; }
    public bool PriorityService { get; set; }
    public bool Insurance { get; set; }
    public int FreeTransfers { get; set; }
    public decimal TravelBonus { get; set; }
    public string ServiceLevel { get; set; }

    public PremiumCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        SupportAssistant = supportAssistant;
        PriorityService = true;
        Insurance = true;
        CardType = "Premium";
        FreeTransfers = 10;
        TravelBonus = 5000;
        ServiceLevel = "VIP";
    }

    public override void CheckBalance()
    {
        if (Balance >= 0)
        {
            Console.WriteLine($"Баланс карты: {Balance} руб.");
        }
        else
        {
            Console.WriteLine("Обнаружена проблема с балансом.");
            Console.WriteLine($"Обратитесь к ассистенту поддержки: {SupportAssistant}.");
        }
    }

    public void RequestPriorityService()
    {
        if (PriorityService)
        {
            Console.WriteLine("Запрошено приоритетное обслуживание.");
        }
        else
        {
            Console.WriteLine("Приоритетное обслуживание недоступно.");
        }
    }

    public void ShowInsurance()
    {
        if (Insurance)
        {
            Console.WriteLine("Страховка карты подключена.");
        }
        else
        {
            Console.WriteLine("Страховка карты отсутствует.");
        }
    }

    public void ChangeAssistant(string newAssistant)
    {
        SupportAssistant = newAssistant;
        Console.WriteLine($"Новый ассистент: {SupportAssistant}");
    }

    public void UseFreeTransfer()
    {
        if (FreeTransfers > 0)
        {
            FreeTransfers--;
            Console.WriteLine($"Бесплатных переводов осталось: {FreeTransfers}");
        }
        else
        {
            Console.WriteLine("Бесплатные переводы закончились.");
        }
    }

    public void ShowServiceLevel()
    {
        Console.WriteLine($"Уровень обслуживания: {ServiceLevel}");
        Console.WriteLine($"Бонус на путешествия: {TravelBonus} руб.");
    }

    public void ExtendInsurance()
    {
        Insurance = true;
        Console.WriteLine("Страховка продлена.");
    }
}

class CorporateCreditCard : CreditCard, IOnlinePayment, ITransfer
{
    public string Company { get; set; }
    public string CompanyPosition { get; set; }
    public decimal CompanyLimit { get; set; }
    public int EmployeeCount { get; set; }
    public string Department { get; set; }
    public decimal MonthlyLimit { get; set; }
    public int TransactionCount { get; set; }

    public CorporateCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, string company)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        Company = company;
        CompanyPosition = "Сотрудник";
        CompanyLimit = 500000;
        EmployeeCount = 1;
        CardType = "Corporate";
        Department = "Общий отдел";
        MonthlyLimit = 1000000;
        TransactionCount = 0;
    }

    public override string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" +
               $"Владелец: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.\n" +
               $"Компания: {Company}\n" +
               $"Должность: {CompanyPosition}\n" +
               $"Лимит компании: {CompanyLimit} руб.\n" +
               $"Количество сотрудников: {EmployeeCount}\n" +
               $"Отдел: {Department}\n" +
               $"Месячный лимит: {MonthlyLimit} руб.\n" +
               $"Количество операций: {TransactionCount}";
    }

    public override void Pay(decimal amount)
    {
        decimal oldBalance = Balance;

        base.Pay(amount);

        if (Balance < oldBalance)
        {
            TransactionCount++;
        }
    }

    public void IncreaseLimit(decimal amount)
    {
        if (amount > 0)
        {
            CompanyLimit += amount;
            Console.WriteLine($"Лимит увеличен до {CompanyLimit} руб.");
        }
    }

    public void ShowCompanyInfo()
    {
        Console.WriteLine($"Компания: {Company}");
        Console.WriteLine($"Должность: {CompanyPosition}");
        Console.WriteLine($"Сотрудников: {EmployeeCount}");
    }

    public void AddEmployee()
    {
        EmployeeCount++;
        Console.WriteLine($"Количество сотрудников: {EmployeeCount}");
    }

    void IOnlinePayment.OnlinePayment(decimal amount)
    {
        Console.WriteLine("Выполнение онлайн-оплаты.");
        Pay(amount);
    }

    void ITransfer.MakeTransfer(CreditCard card, decimal amount)
    {
        Transfer(card, amount);
    }

    public void ChangeDepartment(string department)
    {
        Department = department;
        Console.WriteLine($"Отдел изменен: {Department}");
    }

    public void ShowMonthlyLimit()
    {
        Console.WriteLine($"Месячный лимит: {MonthlyLimit} руб.");
    }

    public void IncreaseMonthlyLimit(decimal amount)
    {
        if (amount > 0)
        {
            MonthlyLimit += amount;
            Console.WriteLine($"Месячный лимит увеличен до {MonthlyLimit} руб.");
        }
    }
}

class PremiumGoldCreditCard : GoldCreditCard
{
    public string PersonalManager { get; set; }
    public decimal SpecialLimit { get; set; }
    public decimal TravelInsurance { get; set; }
    public int PersonalOffers { get; set; }
    public string PrivilegeLevel { get; set; }

    public PremiumGoldCreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, int bonusMiles, string personalManager)
        : base(cardNumber, holderName, expiryDate, balance, bonusMiles)
    {
        PersonalManager = personalManager;
        SpecialLimit = 1000000;
        CardType = "Premium Gold";
        TravelInsurance = 5000000;
        PersonalOffers = 5;
        PrivilegeLevel = "Premium Gold VIP";
    }

    public void ShowManager()
    {
        Console.WriteLine($"Персональный менеджер: {PersonalManager}");
    }

    public void IncreaseSpecialLimit(decimal amount)
    {
        if (amount > 0)
        {
            SpecialLimit += amount;
            Console.WriteLine($"Специальный лимит: {SpecialLimit} руб.");
        }
    }

    public void ContactManager()
    {
        Console.WriteLine($"Связь с персональным менеджером: {PersonalManager}");
    }

    public void ShowPrivileges()
    {
        Console.WriteLine($"Уровень привилегий: {PrivilegeLevel}");
        Console.WriteLine($"Страхование путешествий: {TravelInsurance} руб.");
        Console.WriteLine($"Персональных предложений: {PersonalOffers}");
    }

    public void UsePersonalOffer()
    {
        if (PersonalOffers > 0)
        {
            PersonalOffers--;
            Console.WriteLine($"Осталось персональных предложений: {PersonalOffers}");
        }
        else
        {
            Console.WriteLine("Персональных предложений больше нет.");
        }
    }

    public void ChangePrivilegeLevel(string level)
    {
        PrivilegeLevel = level;
        Console.WriteLine($"Новый уровень привилегий: {PrivilegeLevel}");
    }
}

class CreditCardCollection<T> where T : CreditCard
{
    private List<T> cards = new List<T>();

    public void AddCard(T card)
    {
        cards.Add(card);
        Console.WriteLine($"Карта владельца {card.HolderName} добавлена.");
    }

    public void ShowCards()
    {
        foreach (T card in cards)
        {
            Console.WriteLine(card.GetInfo());
            Console.WriteLine("--------------------");
        }
    }

    public void MakePayment(T card, decimal amount)
    {
        card.Pay(amount);
    }

    public int GetCardCount()
    {
        return cards.Count;
    }
}

class PaymentService
{
    private IPayment payment;
    private ICardInfo cardInfo;

    public PaymentService(IPayment payment, ICardInfo cardInfo)
    {
        this.payment = payment;
        this.cardInfo = cardInfo;
    }

    public void ShowCardInfo()
    {
        Console.WriteLine(cardInfo.GetInfo());
    }

    public void MakePayment(decimal amount)
    {
        Console.WriteLine("Выполнение платежа через PaymentService.");
        payment.Pay(amount);
    }
}

CreditCard card = new CreditCard("1111-2222-3333-4444", "Иван Иванов", "28.12.2036", 50000);

GoldCreditCard goldCard = new GoldCreditCard("2222-3333-4444-5555", "Петр Петров", "29.06.2030", 100000, 500);

PremiumCreditCard premiumCard = new PremiumCreditCard("3333-4444-5555-6666", "Анна Смирнова", "30.07.2029", 75000, "Анна");

CorporateCreditCard corporateCard = new CorporateCreditCard("4444-5555-6666-7777", "Сергей Сидоров", "18.09.2028", 200000, "ООО Ромашка");

PremiumGoldCreditCard premiumGoldCard = new PremiumGoldCreditCard("5555-6666-7777-8888", "Мария Петрова", "15.05.2032", 150000, 1000, "Елена");

Console.WriteLine("=== Обычная кредитная карта ===");
Console.WriteLine(card.GetInfo());
card.CheckBalance();
card.ChangeBank("Сбербанк");
card.ChangePaymentSystem("Visa");
card.ChangeDailyLimit(150000);
card.ShowLastOperation();

Console.WriteLine();

Console.WriteLine("=== Золотая кредитная карта ===");
Console.WriteLine(goldCard.GetInfo());
goldCard.Pay(10000);
goldCard.ShowBonusInfo();
Console.WriteLine($"Кэшбэк с 10000 руб.: {goldCard.CalculateCashback(10000)} руб.");
goldCard.AddBonusPoints(100);
goldCard.ShowBonusProgram();
goldCard.UpgradeBonusLevel();

Console.WriteLine();

Console.WriteLine("=== Премиальная кредитная карта ===");
Console.WriteLine(premiumCard.GetInfo());
premiumCard.CheckBalance();
premiumCard.RequestPriorityService();
premiumCard.ShowInsurance();
premiumCard.UseFreeTransfer();
premiumCard.ShowServiceLevel();

Console.WriteLine();

Console.WriteLine("=== Корпоративная кредитная карта ===");
Console.WriteLine(corporateCard.GetInfo());
corporateCard.Pay(15000);
corporateCard.ShowCompanyInfo();
corporateCard.IncreaseLimit(100000);
corporateCard.AddEmployee();
corporateCard.ChangeDepartment("Бухгалтерия");
corporateCard.ShowMonthlyLimit();
corporateCard.IncreaseMonthlyLimit(200000);

Console.WriteLine();

Console.WriteLine("=== Premium Gold ===");
Console.WriteLine(premiumGoldCard.GetInfo());
premiumGoldCard.ShowBonusInfo();
premiumGoldCard.ShowManager();
premiumGoldCard.IncreaseSpecialLimit(200000);
premiumGoldCard.ShowPrivileges();
premiumGoldCard.UsePersonalOffer();

Console.WriteLine();

Console.WriteLine("=== Перегрузка методов Pay() ===");
card.Pay(1000);
card.Pay(2000, "Покупка продуктов");
card.Pay(3000, "Покупка одежды", "Покупки");

Console.WriteLine();

Console.WriteLine("=== Явная реализация IPayment ===");
IPayment payment = goldCard;
payment.Pay(1000);

Console.WriteLine();

Console.WriteLine("=== Явная реализация ICardInfo ===");
ICardInfo cardInfo = premiumCard;
Console.WriteLine(cardInfo.GetInfo());

Console.WriteLine();

Console.WriteLine("=== Явная реализация IOnlinePayment ===");
IOnlinePayment onlinePayment = corporateCard;
onlinePayment.OnlinePayment(5000);

Console.WriteLine();

Console.WriteLine("=== Явная реализация ITransfer ===");
ITransfer transfer = corporateCard;
transfer.MakeTransfer(premiumCard, 3000);

Console.WriteLine();

Console.WriteLine("=== Перевод между картами ===");
Console.WriteLine($"Баланс Петра до перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны до перевода: {premiumCard.Balance} руб.");

goldCard.Transfer(premiumCard, 5000);

Console.WriteLine($"Баланс Петра после перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны после перевода: {premiumCard.Balance} руб.");

Console.WriteLine();

Console.WriteLine("=== Обобщенная коллекция кредитных карт ===");

CreditCardCollection<CreditCard> collection = new CreditCardCollection<CreditCard>();

collection.AddCard(card);
collection.AddCard(goldCard);
collection.AddCard(premiumCard);
collection.AddCard(corporateCard);
collection.AddCard(premiumGoldCard);

Console.WriteLine($"Всего карт: {collection.GetCardCount()}");
collection.ShowCards();

Console.WriteLine();

Console.WriteLine("=== Полиморфизм ===");

CreditCard selectedCard;

selectedCard = goldCard;
selectedCard.Pay(1000);

selectedCard = premiumCard;
selectedCard.CheckBalance();

selectedCard = corporateCard;
Console.WriteLine(selectedCard.GetInfo());

selectedCard = premiumGoldCard;
selectedCard.Pay(2000);

Console.WriteLine();

Console.WriteLine("=== Управление зависимостями ===");

PaymentService goldPaymentService = new PaymentService(goldCard, goldCard);

Console.WriteLine("Информация о золотой карте:");
goldPaymentService.ShowCardInfo();
goldPaymentService.MakePayment(2000);

Console.WriteLine();

PaymentService corporatePaymentService = new PaymentService(corporateCard, corporateCard);

Console.WriteLine("Информация о корпоративной карте:");
corporatePaymentService.ShowCardInfo();
corporatePaymentService.MakePayment(3000);
```
