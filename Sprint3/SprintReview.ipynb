Создайте программу, демонстрирующую принципы полиморфизма, перегрузки и перекрытия методов в контексте класса "Компьютер" и его компонентов. Используйте обобщенные типы (Generic) для создания гибкой системы.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

interface IInstallable
{
    void Install();
}

abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} — {Price} руб.";
    }
}

class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string name, decimal price, int cores, double frequency)
        : base(name, price)
    {
        Cores = cores;
        Frequency = frequency;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}, ядер: {Cores}, частота: {Frequency} ГГц, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (ReferenceEquals(a, null) || ReferenceEquals(b, null))
            return false;

        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}
class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public int Speed { get; set; }

    public RAM(string name, decimal price, int capacity, int speed)
        : base(name, price)
    {
        Capacity = capacity;
        Speed = speed;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память: {Name}, объём: {Capacity} ГБ, скорость: {Speed} МГц, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Оперативная память {Name} установлена.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (ReferenceEquals(a, null) || ReferenceEquals(b, null))
            return false;

        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

class GPU : Component, IInstallable
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory)
        : base(name, price)
    {
        Memory = memory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}, видеопамять: {Memory} ГБ, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (ReferenceEquals(a, null) || ReferenceEquals(b, null))
            return false;

        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public int RotationSpeed { get; set; }

    public HDD(string name, decimal price, int capacity, int rotationSpeed)
        : base(name, price)
    {
        Capacity = capacity;
        RotationSpeed = rotationSpeed;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Жёсткий диск: {Name}, объём: {Capacity} ГБ, скорость вращения: {RotationSpeed} об/мин, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Жёсткий диск {Name} установлен.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (ReferenceEquals(a, null) || ReferenceEquals(b, null))
            return false;

        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

class Computer
{
    private List<Component> components = new List<Component>();


    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();

        Console.WriteLine($"Компонент {component.Name} добавлен в компьютер.\n");
    }

    public void AddComponent<T>(T component, bool install)
        where T : Component, IInstallable
    {
        components.Add(component);

        if (install)
        {
            component.Install();
        }

        Console.WriteLine($"Компонент {component.Name} добавлен в компьютер.\n");
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удалён из компьютера.\n");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден.\n");
        }
    }

    public void ShowComponents()
    {
        Console.WriteLine("=== Компоненты компьютера ===");

        foreach (Component component in components)
        {
            component.GetInfo();
        }

        Console.WriteLine();
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;

        foreach (Component component in components)
        {
            total += component.Price;
        }

        return total;
    }
}

static class ComputerFactory
{

    public static Computer CreateOfficeComputer<T1, T2>(T1 cpu, T2 ram)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
    {
        Computer computer = new Computer();

        computer.AddComponent(cpu);
        computer.AddComponent(ram);

        return computer;
    }

    public static Computer CreateGamingComputer<T1, T2, T3, T4>(
        T1 cpu, T2 ram, T3 gpu, T4 hdd)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
        where T3 : Component, IInstallable
        where T4 : Component, IInstallable
    {
        Computer computer = new Computer();

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);

        return computer;
    }
}



CPU cpu = new CPU("Intel Core i5", 18000m, 6, 3.5);
RAM ram = new RAM("Kingston Fury", 6500m, 16, 3200);
GPU gpu = new GPU("NVIDIA RTX 4060", 35000m, 8);
HDD hdd = new HDD("Seagate Barracuda", 4500m, 2000, 7200);

Computer computer = new Computer();

Console.WriteLine("=== Добавление компонентов ===");

computer.AddComponent(cpu);
computer.AddComponent(ram, false);
computer.AddComponent(gpu);
computer.AddComponent(hdd);

Console.WriteLine("=== Информация о компьютере ===");
computer.ShowComponents();

Console.WriteLine($"Общая стоимость: {computer.GetTotalPrice()} руб.\n");

Console.WriteLine("=== Удаление компонента ===");
computer.RemoveComponent(hdd);

Console.WriteLine($"Стоимость после удаления: {computer.GetTotalPrice()} руб.\n");

Console.WriteLine("=== Проверка операторов сравнения ===");

CPU anotherCpu = new CPU("Intel Core i5", 18000m, 6, 3.5);

Console.WriteLine($"Процессоры равны: {cpu == anotherCpu}");
Console.WriteLine($"Процессоры не равны: {cpu != anotherCpu}");


Console.WriteLine("\n=== Создание офисного компьютера ===");

Computer officeComputer = ComputerFactory.CreateOfficeComputer(
    new CPU("Intel Core i3", 12000m, 4, 3.0),
    new RAM("Kingston", 4000m, 8, 2666)
);

officeComputer.ShowComponents();
Console.WriteLine($"Стоимость офисного компьютера: {officeComputer.GetTotalPrice()} руб.\n");

Console.WriteLine("=== Создание игрового компьютера ===");

Computer gamingComputer = ComputerFactory.CreateGamingComputer(
    cpu, ram, gpu, hdd
);

gamingComputer.ShowComponents();
Console.WriteLine($"Стоимость игрового компьютера: {gamingComputer.GetTotalPrice()} руб.");
```
