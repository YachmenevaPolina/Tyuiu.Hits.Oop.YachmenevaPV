Создать базовый класс CreditCard в C#, который будет представлять информацию
о кредитных картах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; set; }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal balance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;
    }

    public virtual string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" +
               $"Холдер: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.";
    }

    public virtual void Pay(decimal amount)
    {
        if (amount <= Balance)
        {
            Balance -= amount;
            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств для оплаты.");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Баланс карты: {Balance} руб.");
    }

    public void Transfer(CreditCard otherCard, decimal amount)
    {
        if (amount <= Balance)
        {
            Balance -= amount;
            otherCard.Balance += amount;

            Console.WriteLine(
                $"{HolderName} перевел {amount} руб. пользователю {otherCard.HolderName}."
            );
        }
        else
        {
            Console.WriteLine("Недостаточно средств для перевода.");
        }
    }
}

class GoldCreditCard : CreditCard
{
    public int BonusMiles { get; set; }

    public GoldCreditCard(
        string cardNumber,
        string holderName,
        string expiryDate,
        decimal balance,
        int bonusMiles)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        BonusMiles = bonusMiles;
    }

    public override void Pay(decimal amount)
    {
        if (amount <= Balance)
        {
            Balance -= amount;

            int miles = (int)(amount / 100);
            BonusMiles += miles;

            Console.WriteLine($"Оплата на сумму {amount} руб. выполнена.");
            Console.WriteLine($"Начислено бонусных миль: {miles}.");
            Console.WriteLine($"Всего бонусных миль: {BonusMiles}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств для оплаты.");
        }
    }
}

class PremiumCreditCard : CreditCard
{
    public string SupportAssistant { get; set; }

    public PremiumCreditCard(
        string cardNumber,
        string holderName,
        string expiryDate,
        decimal balance,
        string supportAssistant)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        SupportAssistant = supportAssistant;
    }

    public override void CheckBalance()
    {
        if (Balance >= 0)
        {
            Console.WriteLine($"Баланс карты: {Balance} руб.");
        }
        else
        {
            Console.WriteLine("Обнаружена проблема с балансом.");
            Console.WriteLine($"Обратитесь к ассистенту поддержки: {SupportAssistant}.");
        }
    }
}

class CorporateCreditCard : CreditCard
{
    public string Company { get; set; }

    public CorporateCreditCard(
        string cardNumber,
        string holderName,
        string expiryDate,
        decimal balance,
        string company)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        Company = company;
    }

    public override string GetInfo()
    {
        return $"Номер карты: {CardNumber}\n" +
               $"Холдер: {HolderName}\n" +
               $"Срок действия: {ExpiryDate}\n" +
               $"Баланс: {Balance} руб.\n" +
               $"Компания: {Company}";
    }
}

CreditCard card = new CreditCard(
    "1111-2222-3333-4444",
    "Иван Иванов",
    "28.12.2036",
    50000
);

GoldCreditCard goldCard = new GoldCreditCard(
    "2222-3333-4444-5555",
    "Петр Петров",
    "29.06.2030",
    100000,
    500
);

PremiumCreditCard premiumCard = new PremiumCreditCard(
    "3333-4444-5555-6666",
    "Анна Смирнова",
    "30.07.2029",
    75000,
    "Анна"
);

CorporateCreditCard corporateCard = new CorporateCreditCard(
    "4444-5555-6666-7777",
    "Сергей Сидоров",
    "18.09.2028",
    200000,
    "ООО Ромашка"
);

Console.WriteLine("Обычная кредитная карта");
Console.WriteLine(card.GetInfo());
card.CheckBalance();

Console.WriteLine();

Console.WriteLine("Золотая кредитная карта");
Console.WriteLine(goldCard.GetInfo());
goldCard.Pay(10000);
goldCard.CheckBalance();

Console.WriteLine();

Console.WriteLine("Премиальная кредитная карта");
Console.WriteLine(premiumCard.GetInfo());
premiumCard.CheckBalance();

Console.WriteLine();

Console.WriteLine("Корпоративная кредитная карта");
Console.WriteLine(corporateCard.GetInfo());
corporateCard.Pay(15000);
corporateCard.CheckBalance();

Console.WriteLine();

Console.WriteLine("Взаимодействие объектов");

Console.WriteLine($"Баланс Петра до перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны до перевода: {premiumCard.Balance} руб.");

goldCard.Transfer(premiumCard, 5000);

Console.WriteLine($"Баланс Петра после перевода: {goldCard.Balance} руб.");
Console.WriteLine($"Баланс Анны после перевода: {premiumCard.Balance} руб.");

Console.WriteLine();

CreditCard selectedCard;

selectedCard = goldCard;
selectedCard.Pay(1000);

selectedCard = premiumCard;
selectedCard.CheckBalance();

selectedCard = corporateCard;
Console.WriteLine(selectedCard.GetInfo());
```
